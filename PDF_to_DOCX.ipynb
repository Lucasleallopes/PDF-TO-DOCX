```python
from pdf2image import convert_from_path
import pytesseract
from docx import Document
from google.colab import files
import os
import re

def sanitize_xml_string(text):
    invalid_xml_chars = re.compile(u'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F-\x84\x86-\x9F]')
    return invalid_xml_chars.sub('', text)

print("Faça o upload do seu arquivo PDF:")
uploaded = files.upload()

if not uploaded:
    print("Upload cancelado. Nenhum arquivo foi enviado.")
else:
    pdf_filename = list(uploaded.keys())[0]
    docx_filename = pdf_filename.replace('.pdf', '_convertido.docx')
    doc = Document()

    print(f"\nConvertendo páginas em imagens (isso pode levar um tempo)...")
    pages = convert_from_path(pdf_filename, dpi=300)

    print("Iniciando a leitura do texto (OCR)...")
    for i, page in enumerate(pages):
        print(f"Extraindo texto da página {i + 1} de {len(pages)}...")
        text = pytesseract.image_to_string(page, lang='por+eng')

        if text.strip():
            cleaned_text = sanitize_xml_string(text)
            doc.add_paragraph(cleaned_text)
            doc.add_page_break()

    doc.save(docx_filename)
    print(f"\nConcluído! Baixando o arquivo {docx_filename}...")
    files.download(docx_filename)

    os.remove(pdf_filename)
    os.remove(docx_filename)
```
